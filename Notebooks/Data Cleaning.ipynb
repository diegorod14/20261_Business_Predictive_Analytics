```python
import pandas as pd

df = pd.read_csv("../Data/demand_forecasting.csv")
df.head()
```

```python
renombrar = {
            "Date":"date",
             "Store ID":"store_id",
             "Product ID" : "product_id",
             "Category":"category",
             "Region": "region",
             "Inventory Level": "inventory_level",
             "Units Sold": "units_sold",
             "Units Ordered": "units_ordered",
             "Price": "price",
             "Discount": "discount",
             "Weather Condition": "weather_condition",
             "Promotion": "is_promotion",
             "Competitor Pricing": "competitor_pricing",
             "Seasonality": "seasonality",
              "Epidemic":  "is_epidemic",
             "Demand" : "demand"
}
df.rename(columns=renombrar, inplace=True)
print(list(df.columns))
```

```python
# Info del dataset
df.info()
```

```python
# Dimensiones del Dataset
print(f"Cantidad de filas: {df.shape[0]}")
print(f"Cantidad de columnas: {df.shape[1]}")
```

```python
# Rango de fechas
print(f"Fecha Inicio: {df['date'].min()}")
print(f"Fecha Fin: {df['date'].max()}")
```

```python
# Verificar valores Unicos
df.nunique().sort_values(ascending=True)
```

```python
# Verificar duplicados
df.duplicated().sum()
print(f"Forma antes de eliminar duplicados: {df.shape}")
df.drop_duplicates(inplace=True)
print(f"Forma despues de eliminar duplicados: {df.shape}")
```

```python
# Verificar nulos
df.isnull().sum()
```

```python
pct_nulos = df.isnull().sum() / len(df) * 100
pct_nulos
```

```python
# Casting de tipos de Datos

df["category"] = df["category"].astype('category')
df["region"] = df["region"].astype('category')
df["is_promotion"] = df["is_promotion"].astype('category')
df["weather_condition"] = df["weather_condition"].astype('category')
df["is_epidemic"] = df["is_epidemic"].astype('category')
df["seasonality"] = df["seasonality"].astype('category')
df['date'] = pd.to_datetime(df['date'])

df.dtypes
```
